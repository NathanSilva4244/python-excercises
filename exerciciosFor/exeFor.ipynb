```python
# Exercício 1 — Números de 1 a 10

for i in range(1, 11):
    print(i)
```

```python
# Exercício 2 — Números de 10 até 1

for i in range(10,0, -1):
    print(i)
```

```python
# Exercício 3 — Apenas números pares

for i in range(0, 21, +2):
    print(i)
```

```python
# Exercício 4 — Tabuada

numero = 9

for i in range(1, 11):
    tabuada = (i * numero)
    print(f'{numero} x {i} = {tabuada}')
```

```python
# Exercício 5 — Triângulo simples

for i in range(0, 5):
    triangulo = '*' * (i + 1)
    print(triangulo)
```

```python
# Exercício 6 — Triângulo invertido

for i in range(6, 0, -1):
    triangulo = '*' * (i - 1)
    print(triangulo)
```

```python
# Exercício 7 — Pirâmide

for i in range(1, 6):
    espaco = ' ' * (6 - i)
    triangulo = '*' * (2 * i - 1)
    
    print(espaco + triangulo)
```
