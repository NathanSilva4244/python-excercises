```python
def triangulo():
    matrix = 6

    for i in range(1, matrix):
        reta = ' ' * (matrix - i)
        triangulo = '*' * (2 * i - 1)
        print(reta + triangulo)

triangulo()

```

```python
def numeros_ordernados():
    numeros = []

    for i in range(1, 6):
        numero = int(input(f'Digite o {i} numero: '))
        numeros.append(numero)
        ordenados = sorted(numeros)
    print(f' {ordenados}')

numeros_ordernados()
```
